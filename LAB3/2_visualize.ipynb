```python
import pandas as pd
from matplotlib import font_manager as fm
df=pd.read_csv('../word_freq.csv')
df.head()
```

```python
font_path = r"C:\Users\evilk\AppData\Local\Microsoft\Windows\Fonts\MANGAL - Copy.ttf"
prop = fm.FontProperties(fname=font_path)
```

```python
sort_df=df.sort_values(by='freq',ascending=False)
sort_df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))
plt.bar(sort_df["word"][:100].astype(str), sort_df["freq"][:100])   # no float() here
plt.xticks(rotation=90, fontsize=10,fontproperties=prop)
plt.xlabel("Word")
plt.ylabel("Frequency")
plt.yscale("log")
plt.title("Top 100 Word Frequencies")
plt.tight_layout()
plt.show()
```

```python
# remove stop words
threshhold=1e6
sort_df.drop(sort_df[sort_df["freq"] < threshhold].index, inplace=True)
```
